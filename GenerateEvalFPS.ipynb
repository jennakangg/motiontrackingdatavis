```python
import json

# Define the color JND and thresholds
color_JND = 1 / 63.333333333
color_thresholds = [
    color_JND * 1.5,
    color_JND * 6,
    color_JND * 8
]
name = "niall"
# Load the JSON data from an existing file
input_file_path = 'trial_ids_fps/trials_fps_2_segments_base.json'  # Replace with the actual file path

with open(input_file_path, 'r') as input_file:
    data = json.load(input_file)

# Loop through each trial and update contrast_intensity based on thresholds
for trial in data:
    for segment in trial['segments']:
        # Map the current contrast_intensity value to the corresponding threshold
        intensity = segment['contrast_intensity']
        if intensity == 8:
            segment['contrast_intensity'] = color_thresholds[2]
        elif intensity == 6:
            segment['contrast_intensity'] = color_thresholds[1]
        elif intensity == 1.5:
            segment['contrast_intensity'] = color_thresholds[0]
        # If intensity is not mapped, it can be left unchanged or handled as needed

# Save the updated data to a new JSON file
output_file_path = f"trial_ids_fps/trials_fps_2_segments_{name}.json"
with open(output_file_path, 'w') as output_file:
    json.dump(data, output_file, indent=4)

print(f"Updated JSON data saved to '{output_file_path}'.")
```

```python
import json
import random

# Function to generate a random sequence of trials

def generate_random_trials(input_file, output_file, repetitions):
    # Load trials from input JSON
    with open(input_file, mode='r') as file:
        trials = json.load(file)

    # Create repeated and shuffled trials
    all_trials = []
    for _ in range(repetitions):
        random.shuffle(trials)
        all_trials.extend(trials)

    # # Assign new trial IDs to preserve uniqueness
    # for idx, trial in enumerate(all_trials):
    #     trial["trial_num"] = idx
    # Save randomized trials to output JSON
    with open(output_file, mode='w') as file:
        json.dump(all_trials, file, indent=4)

    print(f"Randomized trials with {repetitions} repetitions saved to {output_file}")
    print(f"Total trials: {len(all_trials)}")

# Parameters

for i in range(0,1):
    input_file = f"trial_ids_fps/trials_fps_2_segments_{name}.json"  # Path to the input file
    output_file = f"trials_fps/generated_trials_{name}_2_segments_{i}.json"  # Path to save the randomized output file
    repetitions = 5  # Number of repetitions

    # Generate the randomized sequence
    generate_random_trials(input_file, output_file, repetitions)
```
